# 01 — Label Conversations

Read each conversation and fill in your labels below, following `guidelines/labeling_guideline.md`. This notebook is intentionally manual — the point of the exercise is doing the labeling yourself, not automating it away.

```
import json
import pandas as pd
from pathlib import Path

conversations = json.loads(Path('../data/conversations.json').read_text())
print(f'{len(conversations)} conversations loaded')
```

## Read a conversation

Re-run this cell with a different index to read the next one.

```
idx = 0
conv = conversations[idx]
print(conv['conversation_id'])
for turn in conv['turns']:
    print(f"{turn['speaker']:>9}: {turn['text']}")
```

## Record your label for this conversation

Fill in the dict below for the conversation you just read, then run the cell to append it to `labels`. Repeat for every conversation.

```
labels = []  # run this cell once at the start of a labeling session
# labels = pd.read_csv('../data/my_labels.csv').to_dict('records')  # or resume a saved session
```

```
labels.append({
    'conversation_id': conv['conversation_id'],
    'primary_intent': 'REPLACE_ME',       # billing_issue / delivery_delay / refund_request /
                                            # account_access / product_complaint / general_inquiry
    'sentiment': 'REPLACE_ME',             # negative / neutral / positive
    'pii_present': False,                  # True/False — confirm against pii_flags.json
    'pii_types': [],                       # e.g. ['email', 'phone']
    'resolution_status': 'REPLACE_ME',     # resolved / unresolved / unclear
    'low_confidence': False,
})
print(f'{len(labels)} conversations labeled so far')
```

## Save your labels

```
pd.DataFrame(labels).to_csv('../data/my_labels.csv', index=False)
print('Saved to data/my_labels.csv')
```

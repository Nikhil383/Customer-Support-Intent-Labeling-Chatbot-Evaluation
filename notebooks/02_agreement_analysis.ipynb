# 02 — Inter-Annotator Agreement

Label a ~20% sample twice (e.g. on two different days, without looking at your first pass) and compare here. This mirrors the guideline's agreement-checking process and the JD's quality-consistency requirement.

```
import pandas as pd
from sklearn.metrics import cohen_kappa_score

pass1 = pd.read_csv('../data/labels_pass1.csv').sort_values('conversation_id').reset_index(drop=True)
pass2 = pd.read_csv('../data/labels_pass2.csv').sort_values('conversation_id').reset_index(drop=True)

assert pass1['conversation_id'].equals(pass2['conversation_id']), \
    'Both passes must cover the exact same conversation_id sample'
```

```
kappa = cohen_kappa_score(pass1['primary_intent'], pass2['primary_intent'])
agreement_rate = (pass1['primary_intent'] == pass2['primary_intent']).mean()

print(f'Conversations compared: {len(pass1)}')
print(f'Raw agreement rate:     {agreement_rate:.2%}')
print(f"Cohen's kappa:          {kappa:.3f}")
```
